# Modification of tuple elements

```python
mytuple
```

```python
temp = list(mytuple)
print(temp)
temp[3] = "grapes"
print(temp)
mytuple = tuple(temp)

mytuple
```

```python
x = ("dolly", "kitti")
y = ("ammu", "deepu")

print(x+y)
print(y+x)

print()

print(x)
print(y)

print()

x += y        # x = x+y         x = ("dolly", "kitti", "ammu", "deepu")
y += x        # y = y+x         y = y + x = ("ammu", "deepu") + ("dolly", "kitti", "ammu", "deepu")

print(x)
print(y)
```

# Unpacking a tuple

```python
std_names = ("deepu", "ammu", "kitti", "dolly")

std1, std2, std3, std4 = std_names

std2
```

```python
std_names = ("deepu", "ammu", "kitti", "dolly", "sai", "krishna", "kishore")

std1, std2, *remaining_std = std_names

print(std2)
print(remaining_std)
```

```python
std_names = ("deepu", "ammu", "kitti", "dolly", "sai", "krishna", "kishore")

first_std, *remaining_std, last_std = std_names

print(first_std)
print(remaining_std)
print(last_std)
```

```python
for x in std_names:
    print(x)
```

```python
for x in range(len(std_names)):
    print(x)
```

```python
for x in range(len(std_names)):
    print(x, std_names[x])
```

```python
fruits = ("apple", "banana", "cherry")
mytuple = fruits * 2

print(mytuple)
```
